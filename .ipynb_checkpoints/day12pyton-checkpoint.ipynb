```python
#math module
import math

#square root
print("square root of 25:", math.sqrt(25))

#factorial
print("factorial of 5:", math.factorial(5))

#value of pi
print("value of pi:", math.pi)

#cosine of 45 degrees
print("cosine of 45 degrees:", math.cos(math.radians(45)))
```

```python
#datetime module

import datetime

#current date and time
now = datetime.datetime.now()
print("current date and time:", now)

#specific date
d = datetime.date(2024,8,27)
print("specific date:", d)

#difference between two dates
delta = now - datetime.datetime(2024,1,1)
print("days since january 1, 2024:", delta.days)

#formatting dates
formatted_date = now.strftime("%Y-%m-%d %H:%M:%S")
print("formatted date and time:", formatted_date)
```

```python
#os module
import os

#current working directory
print("current working directory:", os.getcwd())

#list files in the current directory
print("files in current directory:", os.listdir('.'))

#create a new directory
os.makedirs('new_directory', exist_ok=True)
print("directory 'new_directory' created.")

#remove the directory
os.rmdir('new_directory')
print("directory 'new_directory' removed.")
```

```python
#sys module
import sys

#print python version
print("python version:", sys.version)

#print system platform
print("platform:", sys.platform)

#print command line arguments
print("command line argument:", sys.argv)
```

```python
#random module
import random

#random integer between 1 and 10
print("random integer between 1 and 10:", random.randint(1, 10))

#random floating-point number between 0 and 1
print("random float between 0 and 1:", random.random())

#random choice from a list
items = ['apple', 'banana', 'cherry']
print("random choice from list:", random.choice(items))
```

```python
def add(x,y):
    """returns the sum of two numbers."""
    return x + y

def subtract(x,y):
    """returns the difference between two numbers."""
    return x-y

def multiply(x,y):
    """returns the product of two numbers."""
    return x* y

def devide(x, y):
    """returns the quotient of two numbers. raises zeroDivisor if the divisor is zero."""
    if y== 0:
        raise valueError("cannot divide by zero.")
    return x / y
```

```python
def main():
    print("simple calculator")
    print("select operation:")
    print("1. Add:")
    print("2. Subtract:")
    print("3. Multiply")
    print("4. Divide")
    #get user choice
    choice = input("enter choice (1/2/3/4):")
    #get user input for numbers
    num1 = float(input("enter first number:"))
    num2 = float(input("enter second number:"))
    #perform the chosen operation
    if choice == '1':
        print(f"{num1} + {num2} = {add(num1, num2)}")
    elif choice == '2':
         print(f"{num1} - {num2} = {subtract(num1, num2)}")
    elif choice == '3':
        print(f"{num1} * {num2} = {multiply(num1, num2)}")
    elif choice == '4':
        print(f"{num1} / {num2} = {divide(num1, num2)}")
    else:
        print("Invalid input! please select a vaild operation.")
```

```python
if __name__ == "__main__":
   main()
```
